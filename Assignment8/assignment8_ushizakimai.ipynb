## 課題1

以下の図を再現してください

- 年齢の分布を、異なる階級幅を使って、4つのサブプロットをわけて描画する
   - 階級幅を適当で決めてよい
- 各サブプロットには、
    - x軸ラベル、y軸ラベル、サブプロット番号(a-d)を指定する
    - 水平方向のグリッド線を追加する
    - ヒストグラムの色を`lightblue`に変更する

![](./Figure1.png)

```python
df.columns
```

```python
age = df["age"].dropna()
```

```python
import matplotlib.pyplot as plt
#日本語フォントを使えるようにする
plt.rcParams['font.family'] = 'Yu Gothic'
```

```python
fig, axes = plt.subplots(2, 2)

axes = axes.flatten()
axes[0].hist(age, bins=5, color="lightblue")
axes[0].set_xlabel("年齢")
axes[0].set_ylabel("度数")
axes[0].set_title("(a)", loc="left")
axes[0].grid(axis="y")

axes[1].hist(age, bins=10, color="lightblue")
axes[1].set_xlabel("年齢")
axes[1].set_ylabel("度数")
axes[1].set_title("(b)", loc="left")
axes[1].grid(axis="y")

axes[2].hist(age, bins=20, color="lightblue")
axes[2].set_xlabel("年齢")
axes[2].set_ylabel("度数")
axes[2].set_title("(c)", loc="left")
axes[2].grid(axis="y")

axes[3].hist(age, bins=30, color="lightblue")
axes[3].set_xlabel("年齢")
axes[3].set_ylabel("度数")
axes[3].set_title("(d)", loc="left")
axes[3].grid(axis="y")

plt.tight_layout()
plt.show()
```

## 課題2

以下の図を再現してください

- 男性乗客と女性乗客の年齢の分布をそれぞれ描画する
- 全乗客の年齢分布を示す灰色ヒストグラムも描画する
- 図のタイトル、x軸ラベル、y軸ラベル、凡例を設定する

![](./Figure2.png)

```python
df["sex"].unique()
```

```python
male_age = df[df["sex"] == "male"]["age"].dropna()
female_age = df[df["sex"] == "female"]["age"].dropna()
```

```python
fig, ax = plt.subplots()

# 全乗客(灰色)
ax.hist(age, bins=10, color="gray", alpha=0.5, label="全乗客")

# 男性
ax.hist(male_age, bins=10, alpha=0.5, label="男性")

# 女性
ax.hist(female_age, bins=10, alpha=0.5, label="女性")

ax.set_title("乗客の年齢分布(性別ごと)")
ax.set_xlabel("年齢")
ax.set_ylabel("度数")
ax.legend()

plt.show()
```
